## 1. Privacidade, Proteção e Publicidade (Encapsulamento)

Em linguagens como C++ (a versão POO do C) ou Java, a memória de um atributo privado é fisicamente bloqueada pelo compilador. O Python, sendo uma linguagem de tipagem dinâmica, não bloqueia a memória. Em vez disso, ele usa convenções e um truque chamado *name mangling* (alteração de nomes). Ou seja, real privacidade não existe em Python o que existe é um "acordo de cavalheiros" entre os programadores que apenas diz:

 "Olha, isso aqui é privado. Evita de mexer, campeão. Que meu código pode crachar e dar muita dor de cabeça para mim e para ti. Mas se for para mecher mesmo, use com responsabilidade que eu não granto que tudo vai sair perfeito." 

As regras básicas ("privacidade", "proteção" e publicidade) do Pyhton para o encapsulamento são:
* **Público:** Totalmente acessível (`self.atributo`).
* **Protegido (1 subtraço):** `self._atributo` indica que o atributo só deve ser acessado pela própria classe e por classes filhas.
* **Privado (2 subtraços):** `self.__atributo` é acessível apenas à própria classe. O Python altera o nome interno desse atributo para evitar que ele seja sobrescrito acidentalmente (ou seja, parte do que é o *name mangling*).

Vamos criar a base da nossa fração usando atributos privados para proteger o numerador e o denominador:

```python
from math import gcd

class Fracao:
    # Construtor com valores padrão, inicializando a fração
    def __init__(self, num: int = 1, den: int = 1):
        # Aqui chamaremos as propriedades que criaremos a seguir
        # Internamente, os dados viverão em self.__num e self.__den
        self.num = num
        self.den = den

    # Método protegido (inicia com _) para uso interno da classe e possíveis filhas
    def _reducao(self, num, den):
        if num != 0 and den != 0:
            mdc = gcd(num, den)
            num = num // mdc
            den = den // mdc
            return num, den
        else:
            return 0, 1
```

## 2. Decoradores, Getters e Setters

Agora, vocês me perguntam:

"Se nossos atributos são privados (`__num`), como o usuário da classe vai interagir com eles sem quebrar o encapsulamento?"

É aqui que entram as funções de acesso.

Em vez de criar métodos tradicionais como `get_num()` (para acessar o valor salvo na memória) e `set_num()` (para modificar o valor salvo na memória), o Python utiliza **Decoradores**. 

Um decorador "envolve" uma função e altera seu comportamento. Ao usarmos `@property` e `@atributo.setter`, criamos um "pseudo-atributo público" que decora um atributo privado. Isso omite os detalhes internos da implementação e simplifica a vida de quem consome a classe.

O *setter* atua como nosso "porteiro", garantindo que a regra de negócio não seja violada (como tentar passar uma *string* no lugar de um inteiro).

```python
# Vamos redefinir nossa classe adicionando os decoradores para que o código seja executável
from math import gcd

class Fracao:
    def __init__(self, num: int = 1, den: int = 1):
        # Agora passamos pelas propriedades para que a validação do setter já atue na criação!
        self.num = num
        self.den = den

    @property
    def num(self):
        return self.__num
    
    @num.setter
    def num(self, valor):
        if isinstance(valor, int):  # Checagem rigorosa de tipo
            self.__num = valor
        else:
            raise TypeError('num deve ser int')
        
    @property
    def den(self):
        return self.__den
    
    @den.setter
    def den(self, valor):
        if isinstance(valor, int):
            self.__den = valor
        else:
            raise TypeError('den deve ser int')

f1 = Fracao(10, 2)
print(f"Numerador acessado via property: {f1.num}")
```

### 💡 Curiosidade 1: Como um Decorador funciona em aplicações reais?

Decoradores não servem apenas para getters/setters. Na minha pesquisa em Machine Learning (ML), por exemplo, uso decoradores para aplicar o Princípio Aberto/Fechado (o "O" do S.O.L.I.D., mais para frente vamos falar nele) no registro de modelos.

Por exemplo, em uma `ClassifierFactory`, eu crio um dicionário vazio e uso um decorador `@ClassifierFactory.register('rf')`. Esse decorador intercepta a função que constrói um *Random Forest* (um tipo de algoritmo de ML), pega essa função e a guarda dentro do dicionário sob a chave `'rf'`. Quando o sistema precisa criar um modelo, para poder treiná-lo em um novo conjunto de dados, ele simplesmente busca no dicionário, sem eu precisar encher o código de blocos `if/else`. A linguagem Python nos permite tratar funções como objetos, e o decorador é a ferramenta perfeita para isso!

*(Para conferir a implementação completa, acessem o repositório da minha pesquisa: [GitHub - RACML (Rainfall Acoustic Classification with Machine Learning)](https://github.com/giovannigremedym/racml)).*

```python
# Snippet isolado e executável baseado no classifiers.py
from typing import Dict, Callable

class ClassifierFactory:
    """ Registry for Machine Learning models via Decorators. """
    _registry: Dict[str, Callable] = {}

    @classmethod
    def register(cls, model_name: str) -> Callable:
        # Decorador que registra o modelo automaticamente no dicionário
        def wrapper(builder_func: Callable) -> Callable:
            cls._registry[model_name.lower()] = builder_func
            return builder_func
        return wrapper

# Aplicando o decorador na prática:
@ClassifierFactory.register('lr')
def build_logistic_regression():
    return "Modelo_Regressao_Logistica_Instanciado"

@ClassifierFactory.register('rf')
def build_random_forest():
    return "Modelo_Random_Forest_Instanciado"
    
print(f"Modelos registrados na Factory magicamente: {list(ClassifierFactory._registry.keys())}")
```

## 3. Dunder Methods: Os "Métodos Privados da Linguagem"

"Se o Python não tem o bloqueio de memória do C++, como a linguagem diferencia um método interno dela mesma de um método criado por nós?"

Eu costumo dizer que os *Dunder Methods* (*Double Underscore*) são, na verdade, os **métodos privados da própria linguagem**. Lembra que para criar um atributo ou método privado nosso nós usamos dois subtraços à esquerda (ex: `__meu_metodo`) e o Python aplica o *name mangling*? Pois bem, para não haver conflito entre os nossos métodos privados e as funções vitais do motor do Python, os criadores da linguagem decidiram colocar os dois subtraços **tanto na esquerda quanto na direita** (ex: `__init__`).

Esses métodos não foram feitos para serem chamados diretamente por você (nós não escrevemos `f1.__add__(f2)`). Eles são gatilhos. Quando você usa um operador matemático, uma função nativa ou roda um script, o Python traduz essa ação para o método privado correspondente da linguagem.

### 3.1. Nascimento, Vida, Morte e Identidade do Objeto

Muito antes de usarmos os dados de uma classe, o Python roda uma verdadeira "linha de montagem" por baixo dos panos usando esses métodos:

*   `__new__(cls)`: É o verdadeiro criador do objeto. Ele aloca o espaço na memória para a instância. Raramente precisamos mexer nele, a não ser que estejamos criando padrões de projeto avançados como o *Singleton* (já já o professor vai abordar ele na disciplina).
*   `__init__(self)`: É o **construtor** / inicializador. Ele roda imediatamente após o `__new__` para definir o estado inicial do objeto (nossos atributos).
*   `__dict__`: Como vimos na curiosidade anterior, é o "cofre" (um dicionário real) onde o Python guarda todas as variáveis da instância.
*   `__del__(self)`: É o **destrutor**. Ele é chamado automaticamente pelo *Garbage Collector* do Python quando o objeto não é mais usado e a memória precisa ser liberada.


```python
class FracaoMemoria:
    # 1. Alocação de memória (raro precisarmos sobrescrever)
    def __new__(cls, *args, **kwargs):
        print("[__new__] Alocando espaço na memória para a Fração...")
        return super().__new__(cls)

    # 2. Inicialização do estado
    def __init__(self, num: int = 1, den: int = 1):
        print(f"[__init__] Inicializando os dados: {num}/{den}")
        self.__num = num
        self.__den = den

    # 3. Limpeza de memória (Destrutor)
    def __del__(self):
        print(f"[__del__] A fração foi deletada e a memória liberada!")

# Mostrando o ciclo de vida na prática
if __name__ == '__main__': # Controle de execução
    print("\n--- Iniciando o Script ---")
    f_temp = FracaoMemoria(5, 2)
    print(f"Dicionário interno: {f_temp.__dict__}")
    
    # Forçando o destrutor (normalmente o Python faz isso sozinho no fim do programa)
    del f_temp 
    print("--- Fim do Script ---\n")
```

### 💡 Curiosidade 3: O Guardião da Execução (Por que o `if __name__ == '__main__':` é tão importante?)

Vocês já devem ter notado que quase todos os nossos códigos de teste, e até os scripts do professor Tarcisio, terminam com o famoso bloco `if __name__ == '__main__':`. Mas o que esses dois *dunders* significam separadamente?

*   **`__name__` (A variável fofoqueira):** Esta é uma variável embutida que **todo módulo Python possui automaticamente**. Ela guarda, na forma de *string*, o nome do arquivo que está sendo executado no momento. Se o seu arquivo se chama `matematica.py`, e você o importar em outro arquivo, a variável `__name__` dentro dele valerá `'matematica'`.
*   **`__main__` (O selo de arquivo principal):** Esta é uma *string* muito especial que o Python usa como selo VIP. O interpretador atribui a *string* `'__main__'` à variável `__name__` **se, e somente se,** o arquivo estiver sendo executado diretamente como o script principal (ou seja, você clicou em "Run" nele ou rodou `python arquivo.py` no terminal). 

**Por que isso importa?**
Isso nos permite usar um arquivo de duas maneiras diferentes. Nós podemos importar nossas classes (como a `Fracao`) em outros projetos sem rodar a bateria de testes que fica no final do arquivo. Aquele bloco `if` diz ao Python: *"Se eu for o script principal sendo rodado, execute este bloco de testes. Mas se alguém estiver apenas me importando como uma biblioteca, ignore estes testes!"*.

```python
# Simulando como o Python lida com __name__ e __main__

def simular_execucao(nome_da_execucao):
    print(f"O Python definiu a variável __name__ como: '{nome_da_execucao}'")
    
    # O nosso clássico teste de execução
    if nome_da_execucao == '__main__':
        print("-> STATUS: Eu sou o script principal! Rodando os testes, abrindo janelas, etc...\n")
    else:
        print(f"-> STATUS: Eu fui apenas importado como o módulo '{nome_da_execucao}'. Ficando quietinho no meu canto.\n")

# Cenário 1: Você deu "Run" diretamente neste arquivo
print("CENÁRIO 1: Execução direta no terminal (python meu_arquivo.py)")
simular_execucao('__main__')

# Cenário 2: Você importou este arquivo em outro projeto (import meu_arquivo)
print("CENÁRIO 2: O arquivo foi importado por outro script (import meu_arquivo)")
simular_execucao('meu_arquivo')

# O valor real desta célula do Jupyter no momento da execução:
print(f"No ambiente real deste Jupyter Notebook, a variável __name__ agora é: {__name__}")
```

### 3.2. Representação Visual e Dimensionamento

"Como o Python sabe o que mostrar na tela quando pedimos para imprimir um objeto complexo?"

Nós ensinamos a ele usando os métodos de representação:

*   `__str__(self)`: É a representação **amigável** para o usuário final, acionada quando usamos `print()` ou `f"{obj}"`.
*   `__repr__(self)`: É a representação **técnica** para o desenvolvedor, muito usada em depuração (debug). Geralmente retorna o código necessário para recriar o objeto (ex: `Fracao(1, 2)`). Por padrão retorna o endereço de memória (todo objeto novo em Pyhton já vem com ele implementado, mas pode ser sobreescrito).
*   `__len__(self)`: Acionado pelo comando `len(obj)`. Deve retornar um inteiro. Faz sentido em classes que são "coleções" ou "containers", como uma equação contendo várias frações.

```python
# Reutilizando a lógica do __str__ de Fracao
class FracaoVisual:
    def __init__(self, num: int, den: int):
        self.num = num
        self.den = den
        
    def __str__(self):
        # Exibição bonita para o usuário
        return f"{self.num}/{self.den}"

    def __repr__(self):
        # Exibição técnica para debug
        return f"FracaoVisual(num={self.num}, den={self.den})"

# Uma classe para demonstrar o dimensionamento (__len__)
class Equacao:
    def __init__(self):
        self.__termos = []
        
    def add(self, fracao):
        self.__termos.append(fracao)
        
    def __len__(self):
        # Medindo a quantidade de frações na equação
        return len(self.__termos)

fv = FracaoVisual(3, 4)
print(f"Print chama o __str__: {fv}")
print(f"O __repr__ mostra os detalhes técnicos: {repr(fv)}")

eq = Equacao()
eq.add(fv)
print(f"Tamanho da Equação (__len__): {len(eq)} termos.")
print(f"O __repr__ mostrando o valor padrão: {repr(eq)}")
```

### 3.3. Sobrescrita vs. Sobrecarga (A Mágica dos Operadores)

Antes de colocarmos a mão no código, precisamos esclarecer dois conceitos da Orientação a Objetos que têm nomes parecidos, mas propósitos diferentes: **Sobrescrita** e **Sobrecarga**.

*   **Sobrescrita (Overriding):** Ocorre quando uma classe filha herda um método de sua classe pai e decide "reescrevê-lo" para mudar seu comportamento. Por exemplo, todas as nossas classes herdam o método `__str__` do objeto primordial `object`. Se não fizermos nada, ele imprime o endereço de memória. Nós *sobrescrevemos* esse método para que ele imprima "num/den".
*   **Sobrecarga (Overloading):** Ocorre quando damos um "significado extra" a uma função ou operador que já existe. O Python já faz isso nativamente: o operador `+` soma se forem números, mas concatena se forem *strings*. Quando implementamos métodos como `__add__` ou `__mul__`, estamos *sobrecarregando* os operadores `+` e `*` para que eles passem a entender como lidar com instâncias da nossa classe `Fracao`.

**Atenção ao Paradigma Funcional:** Um método matemático sobrecarregado nunca deve alterar os atributos do próprio objeto que chamou a operação. Ele deve calcular o resultado e **retornar uma nova instância** da classe.

### 3.3. Operadores Matemáticos, Lógicos e Relacionais

É aqui que a mágica (o Polimorfismo) acontece. Nós não precisamos criar métodos engessados como `f1.somar(f2)` ou `f1.eh_igual(f2)`. Nós podemos usar `f1 + f2` e `f1 == f2`. O Python intercepta os sinais e chama os métodos dunder correspondentes.

Vocês mapearam todos eles no Guia de Referência da AP1! Aqui está a tradução da linguagem matemática para a "linguagem privada" do Python:

**Operadores Matemáticos:**
*   Soma (`+`): `__add__`
*   Subtração (`-`): `__sub__`
*   Multiplicação (`*`): `__mul__`
*   Potência (`**`): `__pow__`
*   Divisão Normal (`/`): `__truediv__` --> Retorna um `float`
*   Divisão Inteira (`//`): `__floordiv__` --> Retorna um `int` (valor arredondado)
*   Módulo / Resto (`%`): `__mod__` --> Retorna o reto da divisão

**Operadores Unários (agem sobre um único objeto):**
*   Positivo (`+obj`): `__pos__`
*   Negativo (`-obj`): `__neg__`
*   Inversão de Bits (`~obj`): `__invert__`

**Operadores Lógicos / Bit a Bit (Bitwise):**
O Python também nos permite sobrecarregar operações que ocorrem diretamente nos "fios de cobre" da máquina: a manipulação de zeros e uns (bits). São os operadores *Bitwise*:
*   AND (`&`): `__and__`
*   OR (`|`): `__or__`
*   XOR (`^`): `__xor__`
*   Shifts (`<<`, `>>`): `__lshift__`, `__rshift__`

**Operadores Relacionais (Comparações):**
*   Menor que (`<`): `__lt__` (Less Than)
*   Maior que (`>`): `__gt__` (Greater Than)
*   Menor ou igual (`<=`): `__le__` (Less or Equal)
*   Maior ou igual (`>=`): `__ge__` (Greater or Equal)
*   Igualdade (`==`): `__eq__` (Equal)
*   Diferente (`!=`): `__ne__` (Not Equal)

Abaixo está o código implementando essas sobrecargas na nossa classe `Fracao`!

**Dica de Engenheiro de Software:** Nós não precisamos fazer contas complexas de Fração em todos eles! Se nós ensinarmos o Python a descobrir se algo é **Igual (==)** e se é **Menor Que (<)**, a lógica matemática pura resolve o resto:
* Se não é Igual, é Diferente (`!=` é `not ==`).
* Menor ou Igual (`<=`) é simplesmente `(<) or (==)`.
* Maior Que (`>`) é simplesmente o inverso de Menor ou Igual (`not <=`).
* Maior ou Igual (`>=`) é o inverso de Menor Que (`not <`).

Basta implementarmos a lógica cruzada para o `__eq__` e o `__lt__`, e o resto a gente deduz!

```python
from math import gcd

class Fracao:
    def __init__(self, num: int = 1, den: int = 1):
        self.num = num
        self.den = den

    def _reducao(self, num, den):
        if num != 0 and den != 0:
            mdc = gcd(num, den)
            return num // mdc, den // mdc
        return 0, 1

    @property
    def num(self): return self.__num
    @num.setter
    def num(self, valor):
        if isinstance(valor, int): self.__num = valor
        else: raise TypeError('num deve ser int')
        
    @property
    def den(self): return self.__den
    @den.setter
    def den(self, valor):
        if isinstance(valor, int): self.__den = valor
        else: raise TypeError('den deve ser int')
        
    # ==========================================
    # 1. OPERADORES MATEMÁTICOS BÁSICOS
    # ==========================================
    def __add__(self, other): # Soma (+)
        if isinstance(other, Fracao):
            num = (self.__num * other.den) + (self.__den * other.num)
            den = self.__den * other.den
            num, den = self._reducao(num, den)
            return Fracao(num, den) # Nova Instância!
        raise TypeError('[__add__] somente frações aceitas')
        
    def __sub__(self, other): # Subtração (-)
        if isinstance(other, Fracao):
            num = (self.__num * other.den) - (self.__den * other.num)
            den = self.__den * other.den
            num, den = self._reducao(num, den)
            return Fracao(num, den)
        raise TypeError('[__sub__] somente frações')
        
    def __mul__(self, other): # Multiplicação (*)
        if isinstance(other, Fracao):
            num = self.__num * other.num
            den = self.__den * other.den
            num, den = self._reducao(num, den)
            return Fracao(num, den)
        raise TypeError('[__mul__] somente frações')
        
    def __truediv__(self, other): # Divisão Real (/)
        if isinstance(other, Fracao):
            if other.num == 0: raise ZeroDivisionError('Divisão por zero!')
            num = self.__num * other.den
            den = self.__den * other.num
            num, den = self._reducao(num, den)
            return Fracao(num, den)
        raise TypeError('[__truediv__] somente frações')
        
    def __floordiv__(self, other): # Divisão Inteira (//)
        # Para frações, usaremos a mesma lógica, mas poderíamos aplicar num // den puro
        return self.__truediv__(other)
        
    def __mod__(self, other): # Resto da Divisão (%)
        if isinstance(other, Fracao):
            # Simplificação didática: resto da divisão cruzada
            num = (self.__num * other.den) % (self.__den * other.num)
            den = self.__den * other.den
            num, den = self._reducao(num, den)
            return Fracao(num, den)
        raise TypeError('[__mod__] somente frações')

    def __pow__(self, other): # Potência (**)
        if isinstance(other, Fracao):
            # Elevando a fração por um número inteiro (simplificação para o other.num)
            return Fracao(self.__num ** other.num, self.__den ** other.num)
        raise TypeError('[__pow__] somente frações')

    # ==========================================
    # 2. OPERADORES UNÁRIOS
    # ==========================================
    def __pos__(self): # (+obj)
        return Fracao(self.__num, self.__den)
        
    def __neg__(self): # (-obj)
        return Fracao(-self.__num, self.__den)

    def __invert__(self): # Inversão de bits (~obj)
        return Fracao(~self.__num, ~self.__den)

    # ==========================================
    # 3. OPERADORES RELACIONAIS
    # ==========================================
    def __eq__(self, other): # Igualdade (==)
        if isinstance(other, Fracao):
            # Multiplicação cruzada para garantir igualdade mesmo sem redução
            return self.__num * other.den == self.__den * other.num
        return False
    
    def __lt__(self, other): # Menor Que (<)
        if isinstance(other, Fracao):
            return self.__num * other.den < self.__den * other.num
        raise TypeError('Comparação inválida')

    def __ne__(self, other): # Diferente (!=)
        return not self.__eq__(other)
        
    def __le__(self, other): # Menor ou Igual (<=)
        return self.__lt__(other) or self.__eq__(other)
        
    def __gt__(self, other): # Maior Que (>)
        return not self.__le__(other)
        
    def __ge__(self, other): # Maior ou Igual (>=)
        return not self.__lt__(other)

    # ==========================================
    # 4. OPERADORES BIT A BIT (BITWISE)
    # ==========================================
    def __and__(self, other): # AND (&)
        if isinstance(other, Fracao):
            return Fracao(self.__num & other.num, self.__den & other.den)
        raise TypeError('Esperado Fracao')

    def __or__(self, other): # OR (|)
        if isinstance(other, Fracao):
            return Fracao(self.__num | other.num, self.__den | other.den)
        raise TypeError('Esperado Fracao')

    def __xor__(self, other): # XOR (^)
        if isinstance(other, Fracao):
            return Fracao(self.__num ^ other.num, self.__den ^ other.den)
        raise TypeError('Esperado Fracao')

    def __lshift__(self, other): # Shift Esquerda (<<)
        if isinstance(other, Fracao):
            return Fracao(self.__num << other.num, self.__den << other.den)
        raise TypeError('Esperado Fracao')

    def __rshift__(self, other): # Shift Direita (>>)
        if isinstance(other, Fracao):
            return Fracao(self.__num >> other.num, self.__den >> other.den)
        raise TypeError('Esperado Fracao')

    # ==========================================
    # 5. REPRESENTAÇÃO
    # ==========================================
    def __str__(self):
        return f'{self.__num}/{self.__den}'

if __name__ == '__main__':
    f1 = Fracao(1, 2)
    f2 = Fracao(1, 4)
    f3 = Fracao(2, 4) # Equivalente a 1/2
    
    print("====== TESTE MATEMÁTICO ======")
    print(f"Soma (+): {f1 + f2}")
    print(f"Subtração (-): {f1 - f2}")
    print(f"Potência (**): {f1 ** Fracao(2, 1)}") # (1/2) ^ 2
    
    print("\n====== TESTE RELACIONAL (Dedução Lógica) ======")
    print(f"f1 == f3 (Mesmo valor, sem reduzir)? {f1 == f3}")
    print(f"f1 != f2 (Diferente)? {f1 != f2}")
    print(f"f1 >= f2 (Maior ou Igual)? {f1 >= f2}")
    print(f"f2 <= f1 (Menor ou Igual)? {f2 <= f1}")
    
    print("\n====== TESTE BITWISE (Didático) ======")
    # f1(1,2) & f2(1,4) -> Num: 1 & 1 = 1 | Den: 2 & 4 = 0
    print(f"AND Binário (f1 & f2): {f1 & f2}")
    print(f"Inversão Unária (~f1): {~f1}")
    
    print("\n====== TESTE UNÁRIO ======")
    print(f"Fração Negativa (-f1): {-f1}")
```

## 4. Tratamento de Exceções: Blindando a Arquitetura

Na Engenharia de Software, o usuário (ou outro programador consumindo sua classe) **vai** cometer erros. Ele vai tentar passar uma letra onde deveria ser um número, ou vai tentar dividir por zero. Se não protegermos nossa arquitetura, o programa vai simplesmente "crachar", exibindo uma tela vermelha e encerrando a execução abruptamente.

Para evitar isso, usamos o mecanismo de **Tratamento de Exceções**. 

*   **O Gatilho (`raise`):** Quando nosso código detecta que uma regra de negócio foi violada (ex: `denominador == 0`), nós "levantamos" um erro propositalmente usando a palavra-chave `raise`. É como acionar um alarme de incêndio.
*   **O Escudo (`try/except`):** Quando usamos um código que pode dar problema, nós o testamos dentro de um bloco `try:`. Se o alarme soar, o código não "morre"; ele desvia imediatamente para o bloco `except Exception as e:`, onde podemos tratar o erro com elegância (salvando em um log, avisando o usuário, etc).

No Guia de Referência da AP1 que vocês mapearam, estão listadas as principais exceções nativas do Python. Aqui está o que cada uma significa no dia a dia:

*   `TypeError`: Ocorre quando uma operação é aplicada a um tipo inadequado (ex: somar `Fracao` com `string`). Nós usamos muito essa!
*   `ZeroDivisionError`: Disparado ao tentar dividir algo por zero (implementamos isso no nosso `__floordiv__`!).
*   `ValueError`: O tipo está certo, mas o valor é inapropriado (ex: `int('texto')` - a função espera *string*, mas não uma que não possa virar número).
*   `RuntimeError`: Um erro genérico de tempo de execução, usado quando o problema não se encaixa em categorias mais específicas.
*   `IndexError`: Tentativa de acessar um índice que não existe em uma lista (ex: acessar a 5ª posição de uma lista de 3 itens).
*   `KeyError`: Tentativa de acessar uma chave que não existe em um dicionário.
*   `NameError`: Disparado ao tentar usar uma variável ou função que não foi definida.
*   `FileNotFoundError`: Ocorre ao tentar abrir um arquivo que não existe no caminho especificado.

Vamos rodar uma bateria de testes na nossa classe `Fracao` usando o bloco `try/except` para provar que nossos `raise` funcionam como um escudo impenetrável!

```python
# Testando a robustez da nossa Arquitetura
if __name__ == '__main__':
    print('='*60)
    print('[ 1 ] CRIAÇÃO DE FRAÇÕES VÁLIDAS')
    f1 = Fracao(1, 2)
    f2 = Fracao(3, 2)
    print(f"Sucesso! -> f1: {f1} | f2: {f2}")
    
    print('\n[ 2 ] OPERAÇÕES COM FRAÇÕES')
    f3 = f1 + f2
    print(f'Soma (f1 + f2) ............: {f3}')
    print(f'Multiplicação (f1 * f2) ...: {f1 * f2}')
    print(f'f1 é menor que f2? ........: {f1 < f2}')
    
    print('\n' + '='*60)
    print('[ 3 ] BATERIA DE ERROS E SEGURANÇA (TRY/EXCEPT)')
    print('='*60)
    
    # Testando TypeError na criação[cite: 3]
    try:
        print("-> Tentando criar: Fracao(0.2, 3)...")
        f_invalida = Fracao(0.2, 3)
    except Exception as e: # Capturando o erro mapeado no ref.py[cite: 3]
        print(f'   [BLOQUEADO] Exceção capturada: {type(e).__name__} - {e}')

    # Testando TypeError na matemática[cite: 3]
    try:
        print("\n-> Tentando somar uma Fracao com um inteiro (f1 + 6)...")
        f_erro = f1 + 6
    except Exception as e:
        print(f'   [BLOQUEADO] Exceção capturada: {type(e).__name__} - {e}')
        
    # Testando ZeroDivisionError na matemática[cite: 3]
    try:
        print("\n-> Tentando forçar uma Divisão por Zero (f1 // Fracao(0, 1))...")
        f_zero = f1 // Fracao(0, 1)
    except Exception as e:
        print(f'   [BLOQUEADO] Exceção capturada: {type(e).__name__} - {e}')
        
    # Testando uma exceção de tipo de dado não coberto pelo nosso operador relacional
    try:
        print("\n-> Tentando comparar Fração com String (f1 < 'texto')...")
        relacional_invalido = f1 < 'texto'
    except Exception as e:
        print(f'   [BLOQUEADO] Exceção capturada: {type(e).__name__} - {e}')

    print('\n' + '='*60)
    print("STATUS FINAL: A classe sobreviveu a todos os testes! Arquitetura 100% segura.")
```

## Dicionário de Termos e Conceitos Base

* **Tipagem Dinâmica:** Em Python, as variáveis não possuem tipo (não precisamos declarar `int a = 3`), apenas os objetos (os dados na memória) é que sabem o seu próprio tipo. A verificação de compatibilidade ocorre em tempo de execução.
* **Name Mangling:** O interpretador Python altera internamente o nome de atributos iniciados com duplo sublinhado (ex: `__atributo` vira `_Classe__atributo`). Isso é feito para evitar conflitos de nomes em heranças e acessos acidentais.
* **Duck Typing:** Um estilo de programação focado em polimorfismo. O tipo exato do objeto é menos importante do que os métodos que ele implementa ("se anda como pato e grasna como pato, então é um pato").
* **Construtor:** Método (ou conjunto de métodos, em Python a dupla `__new__` e `__init__`) responsável por instanciar o objeto na memória e inicializar seu estado com valores padrão ou parametrizados.
* **Destrutor:** Método (`__del__`) acionado nos últimos suspiros de vida de um objeto. Serve para garantir uma limpeza segura, como fechar uma conexão de banco de dados ou liberar recursos de rede, antes que o objeto seja obliterado da memória.
* **Garbage Collector (Coletor de Lixo):** O mecanismo automático do Python que gerencia a memória do seu computador. Ele rastreia e conta quantas variáveis (referências) apontam para um objeto. Quando esse contador chega a zero (ou seja, ninguém mais no seu código está usando aquele dado), o Coletor de Lixo entra em ação silenciosamente, chama o destrutor (`__del__`) e libera a memória RAM para outros processos.
* **Singleton:** Um padrão de projeto (*Design Pattern*) clássico. Ele é usado quando você precisa garantir que exista **apenas uma única instância** de uma classe rodando em todo o seu programa. No Python, implementamos o Singleton interceptando o construtor `__new__` para que, se o objeto já existir, ele retorne a mesma instância em vez de criar uma nova.
